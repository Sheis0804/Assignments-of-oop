**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


student = Student("Ali", 101)
print(student.get_name())
print(student.get_roll_number())

student.set_name("Ahmed")
student.set_roll_number(102)
print(student.get_name())
print(student.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)
account.deposit(500)
account.withdraw(300)
print("Balance:", account.get_balance())
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
class Employee:
    def __init__(self, salary):
        self.__salary = 0
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary


employee = Employee(50000)
print("Salary:", employee.get_salary())
employee.set_salary(-1000)
print("Salary:", employee.get_salary())
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
class Example:
    def __init__(self):
        self._protected_value = "Protected: intended for class/subclass use"
        self.__private_value = "Private: name-mangled inside the class"

    def show_values(self):
        print(self._protected_value)
        print(self.__private_value)


class Child(Example):
    def show_protected(self):
        print(self._protected_value)
        # self.__private_value cannot be accessed by this name in the child class.


obj = Child()
obj.show_values()
obj.show_protected()
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)
print("Balance through method:", account.get_balance())

try:
    print(account.__balance)
except AttributeError as error:
    print("Error:", error)

# Encapsulation protects internal data and lets the class validate changes
# through methods rather than allowing unrestricted direct access.
```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model


car = Car("Toyota", "Corolla")
print("Brand:", car.brand)
print("Model:", car.model)
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
class Animal:
    def eat(self):
        print("Animal eats")


class Mammal(Animal):
    def walk(self):
        print("Mammal walks")


class Dog(Mammal):
    def bark(self):
        print("Dog barks")


dog = Dog()
dog.eat()   # inherited from Animal
dog.walk()  # inherited from Mammal
dog.bark()  # defined in Dog
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
class Shape:
    def describe(self):
        print("This is a shape")


class Circle(Shape):
    def area(self, radius):
        return 3.14159 * radius ** 2


class Square(Shape):
    def area(self, side):
        return side ** 2


circle = Circle()
square = Square()
circle.describe()
print("Circle area:", circle.area(3))
square.describe()
print("Square area:", square.area(4))
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
class Dog:
    def sound(self):
        print("Woof!")


class Cat:
    def sound(self):
        print("Meow!")


class Cow:
    def sound(self):
        print("Moo!")


def make_sound(animal_object):
    animal_object.sound()


for animal in (Dog(), Cat(), Cow()):
    make_sound(animal)
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Point({self.x}, {self.y})"


p1 = Point(2, 3)
p2 = Point(4, 5)
print(p1 + p2)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Python does not support traditional method overloading by defining
# several methods with the same name; the last definition replaces earlier ones.
# A common alternative is to use default arguments.

class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c


calc = Calculator()
print(calc.add(2))
print(calc.add(2, 3))
print(calc.add(2, 3, 4))
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


try:
    computer = Computer()
except TypeError as error:
    print("Cannot create Computer object:", error)

# An abstract class with an unimplemented abstract method cannot be instantiated.
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data")


class Desktop(Computer):
    def process(self):
        print("Desktop is processing data")


laptop = Laptop()
desktop = Desktop()
laptop.process()
desktop.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod


class Car(ABC):
    def start_engine(self):
        print("Engine started")

    @abstractmethod
    def drive(self):
        pass


class Sedan(Car):
    def drive(self):
        print("Sedan is driving")


car = Sedan()
car.start_engine()
car.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using credit card.")


class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using UPI.")


def checkout(payment_method, amount):
    payment_method.pay(amount)


checkout(CreditCardPayment(), 2500)
checkout(UPIPayment(), 1200)
```
